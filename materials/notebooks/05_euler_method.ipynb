# Forward Euler Method

The **Forward Euler method** is the simplest numerical scheme for solving ordinary differential equations (ODEs).

Given an initial value problem:

$$\frac{dy}{dt} = f(t, y), \quad y(t_0) = y_0$$

The Euler update rule is:

$$y_{n+1} = y_n + h \cdot f(t_n, y_n)$$

where $h$ is the step size.

```python
import numpy as np
import matplotlib.pyplot as plt

def euler_method(f, y0, t_span, h):
    """Forward Euler method for solving dy/dt = f(t, y)."""
    t_start, t_end = t_span
    t = np.arange(t_start, t_end + h, h)
    y = np.zeros((len(t), len(y0)))
    y[0] = y0
    for i in range(len(t) - 1):
        y[i + 1] = y[i] + h * np.array(f(t[i], y[i]))
    return t, y
```

## Example: Exponential Decay

The simplest ODE to test our solver:

$$\frac{dy}{dt} = -y, \quad y(0) = 1$$

The exact solution is $y(t) = e^{-t}$.

```python
# Exponential decay: dy/dt = -y
f_decay = lambda t, y: [-y[0]]

t, y = euler_method(f_decay, [1.0], (0, 5), h=0.5)
t_exact = np.linspace(0, 5, 200)
y_exact = np.exp(-t_exact)

plt.figure(figsize=(8, 4))
plt.plot(t_exact, y_exact, 'k-', label='Exact: $e^{-t}$', linewidth=2)
plt.plot(t, y[:, 0], 'o--', color='#61afef', label=f'Euler (h={0.5})', markersize=6)
plt.xlabel('t'); plt.ylabel('y(t)')
plt.title('Forward Euler vs Exact Solution')
plt.legend(); plt.grid(alpha=0.3)
plt.show()
```

## Step Size Matters

The Euler method's accuracy depends on the step size $h$. Smaller steps = better accuracy but more computation.

```python
plt.figure(figsize=(8, 4))
plt.plot(t_exact, y_exact, 'k-', label='Exact', linewidth=2)

colors = ['#e06c75', '#e5c07b', '#98c379', '#61afef']
for h, color in zip([1.0, 0.5, 0.25, 0.1], colors):
    t_h, y_h = euler_method(f_decay, [1.0], (0, 5), h=h)
    plt.plot(t_h, y_h[:, 0], 'o--', color=color, label=f'h={h}', markersize=4)

plt.xlabel('t'); plt.ylabel('y(t)')
plt.title('Effect of Step Size on Euler Method Accuracy')
plt.legend(); plt.grid(alpha=0.3)
plt.show()
```

## Predator-Prey (Lotka-Volterra)

A more interesting system — two coupled ODEs modeling predator-prey dynamics:

$$\frac{dx}{dt} = \alpha x - \beta x y$$
$$\frac{dy}{dt} = \delta x y - \gamma y$$

where $x$ = prey population, $y$ = predator population.

```python
# Lotka-Volterra parameters
alpha, beta, delta, gamma = 1.1, 0.4, 0.1, 0.4

def lotka_volterra(t, state):
    x, y = state
    dxdt = alpha * x - beta * x * y
    dydt = delta * x * y - gamma * y
    return [dxdt, dydt]

t, sol = euler_method(lotka_volterra, [10.0, 5.0], (0, 50), h=0.01)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

ax1.plot(t, sol[:, 0], color='#61afef', label='Prey')
ax1.plot(t, sol[:, 1], color='#e06c75', label='Predators')
ax1.set_xlabel('Time'); ax1.set_ylabel('Population')
ax1.set_title('Population Over Time')
ax1.legend(); ax1.grid(alpha=0.3)

ax2.plot(sol[:, 0], sol[:, 1], color='#c678dd', linewidth=0.5)
ax2.set_xlabel('Prey'); ax2.set_ylabel('Predators')
ax2.set_title('Phase Portrait')
ax2.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
